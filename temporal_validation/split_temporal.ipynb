```python
temporal_splits = {
    0: [
        "CycleGAN",
        "Cascaded Refinement Networks",
        "ProGAN",
        "StarGAN"
    ],
    1: [
        "SN-PatchGAN",
        "BigGAN",
        "IMLE",
        "StyleGAN1"
    ],
    2: [
        "GauGAN",
        "StyleGAN2",
        "DDPM",
        "CIPS"
    ],
    3: [
        "VQGAN",
        "GANformer",
        "ADM",
        "StyleGAN3"
    ],
    4: [
        "LaMa",
        "FaceSynthetics",
        "ProjectedGAN",
        "Palette"
    ],
    5: [
        "VQ-Diffusion",
        "Denoising Diffusion GAN",
        "Glide",
        "Latent Diffusion"
    ],
    6: [
        "Midjourney",
        "MAT",
        "Diffusion GAN (ProjectedGAN)",
        "Diffusion GAN (StyleGAN2)"
    ],
    7: [
        "Stable Diffusion 1.4",
        "Stable Diffusion 1.5",
        "Stable Diffusion 2.1",
        "DeepFloyd IF"
    ],
    8: [
        "Stable Diffusion XL 1.0",
        "DALL-E 3",
        "FLUX 1 Dev",
        "FLUX 1 Schnell"
    ]
}

generator_to_split = {
    gen: w for w, gen_list in temporal_splits.items() for gen in gen_list
}

val_fake["w_split"] = val_fake['generator'].map(generator_to_split)

unmatched =val_fake[val_fake["w_split"].isna()]
if len(unmatched) > 0:
    print("Unmatched generator names:\n", unmatched["generator"].unique())

val_fake.to_csv("fake_validation_temporal_splits.csv", index=False)

```

```python
train_fake["w_split"] = train_fake['generator'].map(generator_to_split)

unmatched =train_fake[train_fake["w_split"].isna()]
if len(unmatched) > 0:
    print("Unmatched generator names:\n", unmatched["generator"].unique())

train_fake.to_csv("fake_train_temporal_splits.csv", index=False)
```
